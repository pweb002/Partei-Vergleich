### Wahlprogramme zur Bundestagswahl
Sammlung der Wahlprogramme zur Bundestagswahl 2025:
[Bundestagswahl Wahlprogramme](https://www.bundestagswahl-bw.de/bundestagswahl-wahlprogramme)

```python
import os
import requests
from tqdm import tqdm
```

```python
def download_pdf_files(download_folder, urls):
    os.makedirs(download_folder, exist_ok=True)
    
    for party, url in tqdm(urls.items(), 
                           desc="Downloading PDFs", 
                           total=len(urls)):
        
        response = requests.get(url, stream=True)
        
        if response.status_code == 200:
            filename = f"{party}_Wahlprogramm.pdf"
            filepath = os.path.join(download_folder, filename)
            
            with open(filepath, "wb") as f:
                for data in response.iter_content(1024): 
                    f.write(data)
        else:
            tqdm.write(f"Error while Downloading PDF: {party} (Status: {response.status_code})")
```

```python
def download_xml_files(download_folder, base_url, start, end):
    os.makedirs(download_folder, exist_ok=True)
    file_extension = ".xml"
    
    for i in tqdm(range(start, end + 1), desc="Downloading XML Files"):
        file_number = f"{i:03}"
        url = base_url + file_number + file_extension
        
        filename = file_number + file_extension
        filepath = os.path.join(download_folder, filename)
        
        response = requests.get(url)
        
        if response.status_code == 200:
            with open(filepath, "wb") as f:
                f.write(response.content)
        else:
            tqdm.write(f"Error while downloading from: {url} (Status: {response.status_code})")
```

```python
download_folder = "./BTP20_Downloads"

base_url = "https://dserver.bundestag.de/btp/20/20"
    
download_xml_files(download_folder, base_url, 1, 208)
```
